```python

print("Daksha.S-1WA24CS101")

import heapq


def parse_input_state(prompt_text):
    print(prompt_text)
    print("Enter 9 numbers (0-8), where 0 represents blank:")

    while True:
        try:
            numbers = list(map(int, input(">> ").split()))

            if len(numbers) != 9:
                print("Error: Enter exactly 9 numbers.")
                continue

            if sorted(numbers) != list(range(9)):
                print("Error: Enter numbers 0 to 8 exactly once.")
                continue

            return tuple(numbers)

        except ValueError:
            print("Error: Enter numbers only.")


def get_neighbors(state):
    neighbors = []
    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move_name in moves:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            pos = nr * 3 + nc
            new_state = list(state)

            new_state[zero], new_state[pos] = (
                new_state[pos], new_state[zero]
            )

            neighbors.append((tuple(new_state), move_name))

    return neighbors


def print_board(state):
    for i in range(0, 9, 3):
        print("   " + " ".join(
            str(x) if x != 0 else "_"
            for x in state[i:i + 3]
        ))
    print()


def is_solvable(start, goal):
    def count_inversions(state):
        nums = [x for x in state if x != 0]
        return sum(
            nums[i] > nums[j]
            for i in range(len(nums))
            for j in range(i + 1, len(nums))
        )

    return count_inversions(start) % 2 == count_inversions(goal) % 2


def heuristic(state, goal):
    # Number of misplaced tiles, excluding blank
    return sum(
        state[i] != goal[i] and state[i] != 0
        for i in range(9)
    )


def astar(initial, goal):
    priority_queue = []
    heapq.heappush(
        priority_queue,
        (heuristic(initial, goal), 0, initial)
    )

    g_cost = {initial: 0}
    parent = {initial: None}
    move_used = {}
    closed = set()

    while priority_queue:
        f, g, state = heapq.heappop(priority_queue)

        if state in closed:
            continue

        closed.add(state)

        if state == goal:
            path = []
            current = state

            while current is not None:
                path.append(
                    (current, move_used.get(current, "Start"))
                )
                current = parent[current]

            path.reverse()

            print("\nA* found the solution!")
            print("Total moves:", len(path) - 1)

            for step, (board, move) in enumerate(path):
                if step == 0:
                    print("\nInitial State:")
                else:
                    print(f"\nStep {step}: Move blank {move}")

                print_board(board)
                print("g(n) =", step)
                print("h(n) =", heuristic(board, goal))
                print(
                    "f(n) =",
                    step + heuristic(board, goal)
                )

            return

        for neighbor, move in get_neighbors(state):
            if neighbor in closed:
                continue

            new_g = g + 1

            if new_g < g_cost.get(neighbor, float("inf")):
                g_cost[neighbor] = new_g
                parent[neighbor] = state
                move_used[neighbor] = move

                h = heuristic(neighbor, goal)
                new_f = new_g + h

                heapq.heappush(
                    priority_queue,
                    (new_f, new_g, neighbor)
                )

    print("No solution found.")


if __name__ == "__main__":
    print("=== 8-Puzzle A* Search Solver ===\n")

    initial_state = parse_input_state(
        "--- Define Initial State ---"
    )

    goal_state = parse_input_state(
        "--- Define Goal State ---"
    )

    if not is_solvable(initial_state, goal_state):
        print("\nInitial state cannot reach the goal state.")
    else:
        astar(initial_state, goal_state)
```
