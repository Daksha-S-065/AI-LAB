```python
print("Daksha.S-1WA24CS101")
import sys
sys.setrecursionlimit(20000)


def parse_input_state(prompt_text):
    print(prompt_text)
    print("Enter 9 numbers (0-8), where 0 represents blank:")

    while True:
        try:
            numbers = list(map(int, input(">> ").split()))

            if len(numbers) != 9:
                print("Error: Enter exactly 9 numbers.")
                continue

            if sorted(numbers) != list(range(9)):
                print("Error: Enter numbers 0 to 8 exactly once.")
                continue

            return (
                tuple(numbers[0:3]),
                tuple(numbers[3:6]),
                tuple(numbers[6:9])
            )

        except ValueError:
            print("Error: Enter numbers only.")


def get_neighbors(board):
    neighbors = []

    br, bc = next(
        (r, c)
        for r in range(3)
        for c in range(3)
        if board[r][c] == 0
    )

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move_name in moves:
        nr, nc = br + dr, bc + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_board = [list(row) for row in board]

            new_board[br][bc], new_board[nr][nc] = (
                new_board[nr][nc],
                new_board[br][bc]
            )

            neighbors.append(
                (tuple(tuple(row) for row in new_board), move_name)
            )

    return neighbors


def print_board(board):
    for row in board:
        print("   " + " ".join(
            str(x) if x != 0 else "_" for x in row
        ))
    print()


def is_solvable(start, goal):
    def inversions(board):
        nums = [n for row in board for n in row if n != 0]
        count = 0

        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] > nums[j]:
                    count += 1

        return count

    return inversions(start) % 2 == inversions(goal) % 2


def dfs_find_all_paths(current_board, goal_board, visited,
                       current_path, all_paths, max_depth):

    if current_board == goal_board:
        all_paths.append(list(current_path))
        return

    if len(current_path) - 1 >= max_depth:
        return

    for neighbor, move in get_neighbors(current_board):
        if neighbor not in visited:
            visited.add(neighbor)
            current_path.append((neighbor, move))

            dfs_find_all_paths(
                neighbor, goal_board, visited,
                current_path, all_paths, max_depth
            )

            current_path.pop()
            visited.remove(neighbor)


if __name__ == "__main__":
    print("=== 8-Puzzle DFS Solver ===\n")

    initial_state = parse_input_state(
        "--- Define Initial State ---"
    )

    goal_state = parse_input_state(
        "--- Define Goal State ---"
    )

    MAX_DEPTH_LIMIT = 3

    if not is_solvable(initial_state, goal_state):
        print("\nInitial state cannot reach the goal state.")

    else:
        print(f"\nRunning DFS up to depth {MAX_DEPTH_LIMIT}...")

        discovered_paths = []

        dfs_find_all_paths(
            initial_state,
            goal_state,
            {initial_state},
            [(initial_state, "Start")],
            discovered_paths,
            MAX_DEPTH_LIMIT
        )

        if discovered_paths:
            print(
                f"\nDFS found {len(discovered_paths)} path(s)."
            )

            for index, path in enumerate(discovered_paths, 1):
                print(f"\n=== DFS PATH #{index} ===")
                print("Total moves:", len(path) - 1)

                for step, (board, move) in enumerate(path):
                    if step == 0:
                        print("Initial State:")
                    else:
                        print(f"Step {step}: Move blank {move}")

                    print_board(board)

        else:
            print("\nNo solution found within depth 3.")
```

```python

print("Daksha.S-1WA24CS101")

import sys
sys.setrecursionlimit(20000)


def parse_input_state(prompt_text):
    print(prompt_text)
    print("Enter 9 numbers (0-8), where 0 represents blank:")

    while True:
        try:
            numbers = list(map(int, input(">> ").split()))

            if len(numbers) != 9:
                print("Error: Enter exactly 9 numbers.")
                continue

            if sorted(numbers) != list(range(9)):
                print("Error: Enter numbers 0 to 8 exactly once.")
                continue

            return (
                tuple(numbers[0:3]),
                tuple(numbers[3:6]),
                tuple(numbers[6:9])
            )

        except ValueError:
            print("Error: Enter numbers only.")


def get_neighbors(board):
    neighbors = []

    br, bc = next(
        (r, c)
        for r in range(3)
        for c in range(3)
        if board[r][c] == 0
    )

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move_name in moves:
        nr, nc = br + dr, bc + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_board = [list(row) for row in board]

            new_board[br][bc], new_board[nr][nc] = (
                new_board[nr][nc],
                new_board[br][bc]
            )

            neighbors.append(
                (tuple(tuple(row) for row in new_board), move_name)
            )

    return neighbors


def print_board(board):
    for row in board:
        print("   " + " ".join(
            str(x) if x != 0 else "_" for x in row
        ))
    print()


def is_solvable(start, goal):
    def inversions(board):
        nums = [n for row in board for n in row if n != 0]
        count = 0

        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] > nums[j]:
                    count += 1

        return count

    return inversions(start) % 2 == inversions(goal) % 2


def dls(current_board, goal_board, visited,
        current_path, all_paths, depth_limit):

    if current_board == goal_board:
        all_paths.append(list(current_path))
        return

    if len(current_path) - 1 >= depth_limit:
        return

    for neighbor, move in get_neighbors(current_board):
        if neighbor not in visited:
            visited.add(neighbor)
            current_path.append((neighbor, move))

            dls(
                neighbor, goal_board, visited,
                current_path, all_paths, depth_limit
            )

            current_path.pop()
            visited.remove(neighbor)


def ids(initial_state, goal_state, max_depth):
    all_paths = []

    # Increase the depth limit iteratively
    for limit in range(max_depth + 1):
        print(f"\nSearching at depth limit: {limit}")

        discovered_paths = []

        dls(
            initial_state,
            goal_state,
            {initial_state},
            [(initial_state, "Start")],
            discovered_paths,
            limit
        )

        if discovered_paths:
            all_paths.extend(discovered_paths)

            # The first depth containing a solution is the shallowest.
            break

    return all_paths


if __name__ == "__main__":
    print("=== 8-Puzzle IDS Solver ===\n")

    initial_state = parse_input_state(
        "--- Define Initial State ---"
    )

    goal_state = parse_input_state(
        "--- Define Goal State ---"
    )

    MAX_DEPTH_LIMIT = 3

    if not is_solvable(initial_state, goal_state):
        print("\nInitial state cannot reach the goal state.")

    else:
        print(f"\nRunning IDS up to depth {MAX_DEPTH_LIMIT}...")

        discovered_paths = ids(
            initial_state,
            goal_state,
            MAX_DEPTH_LIMIT
        )

        if discovered_paths:
            print(
                f"\nIDS found {len(discovered_paths)} path(s)."
            )

            for index, path in enumerate(discovered_paths, 1):
                print(f"\n=== IDS PATH #{index} ===")
                print("Total moves:", len(path) - 1)

                for step, (board, move) in enumerate(path):
                    if step == 0:
                        print("Initial State:")
                    else:
                        print(f"Step {step}: Move blank {move}")

                    print_board(board)

        else:
            print("\nNo solution found within depth 3.")
```
