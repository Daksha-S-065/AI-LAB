```python

print("Daksha S")
print("1WA24CS101")
from itertools import product

def implies(a, b):
    return not a or b

def evaluate(expr, values):
    expr = expr.replace(" ", "")

    if "->" in expr:
        a, b = expr.split("->", 1)
        return implies(evaluate(a, values), evaluate(b, values))

    if "|" in expr:
        a, b = expr.split("|", 1)
        return evaluate(a, values) or evaluate(b, values)

    if "&" in expr:
        a, b = expr.split("&", 1)
        return evaluate(a, values) and evaluate(b, values)

    if expr.startswith("~"):
        return not evaluate(expr[1:], values)

    if expr.startswith("(") and expr.endswith(")"):
        return evaluate(expr[1:-1], values)

    if expr in values:
        return values[expr]

    raise ValueError("Invalid expression: " + expr)


def solve():
    n = int(input("Enter number of propositions: "))
    symbols = input("Enter proposition names: ").split()

    m = int(input("Enter number of KB conditions: "))
    kb = []

    for i in range(m):
        kb.append(input("Enter KB condition " + str(i + 1) + ": "))

    print("\nTRUTH TABLE")
    print("-" * 40)

    print(" ".join(symbols), "| KB")

    models = []

    # Print truth table first
    for combination in product([True, False], repeat=n):
        values = dict(zip(symbols, combination))

        results = [evaluate(sentence, values) for sentence in kb]
        KB = all(results)

        row = " ".join(
            "T" if values[s] else "F" for s in symbols
        )

        print(row, "|", "T" if KB else "F")

        if KB:
            models.append(values)

    # Ask for query AFTER printing truth table
    print("\nEnter query to check entailment:")
    query = input()

    print("\nENTAILMENT RESULT")
    print("-" * 40)

    if not models:
        print("KB is inconsistent.")
        print("KB entails the query: TRUE")
        return

    entailment = all(evaluate(query, model) for model in models)

    if entailment:
        print("KB entails the query: TRUE")
    else:
        print("KB entails the query: FALSE")


if __name__ == "__main__":
    solve()
```
